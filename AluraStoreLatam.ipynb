##¿DataFrames?
📚 Primero: ¿Cómo funcionan los DataFrames?
Imagina que un DataFrame es como una caja organizada de datos:

* Cada columna tiene un nombre (por ejemplo: "Nombre", "Edad", "Salario", etc.).

* Cada fila representa un registro o una observación (por ejemplo: una persona, una venta, un producto, etc.).

* Puedes acceder, modificar, analizar, sumar, contar, etc. a partir de esa estructura.

* Pandas es como el "gerente" que te da herramientas para manipular esa caja de datos de forma súper eficiente.

##¿Cómo organiza la información un data frame?


```
import pandas as pd

# Creamos un DataFrame de ejemplo
data = {
    'Nombre': ['Ana', 'Luis', 'Marta'],
    'Edad': [23, 30, 27],
    'Salario': [3000, 4500, 3800]
}

df = pd.DataFrame(data)

# Sumar la columna 'Edad'
suma_edades = df['Edad'].sum()

# Sumar la columna 'Salario'
suma_salarios = df['Salario'].sum()

print("Suma de edades:", suma_edades)
print("Suma de salarios:", suma_salarios)

```
Esto quiere decir que básicamente un dataframe es un diccionario donde el titulo de la columna es la etiqueta de ubicación y el valor asociado son listas.

✅ En un diccionario de Python común:

* La clave → sería como el nombre de la columna.

* El valor asociado → sería una lista con los datos de esa columna.

ES CÓMO PONER UN DICCIONARIO EN VERTICAL

✅ Cuando pandas recibe eso, lo convierte en un DataFrame:

* Cada clave se vuelve el título de una columna.

* Cada lista se organiza verticalmente (cada elemento de la lista se convierte en una fila en esa columna).


Entonces los indices que van desde cero hasta el número de datos solicitados, las listas van organizadas hacia abajo en cada columna, cómo tal cada fila representa una posición de lista porque se cuentan hacia abajo.

##Aplicando lo aprendido:
Un archivo como .csv, .xlsx o .json es simplemente texto estructurado, NO es un DataFrame aún. al igual que en general un diccionario o estructuras símilares, por ejemplo:



````python
'''
FUNCIÓN PARA SUMAR LAS VENTAS DE CADA LISTA PERTENECIENTE A LA LLAVE PRECIO DE UNA TIENDA
'''
def sumatoria(a):
 return round(a["Precio"].sum(),0)




'''
FUNCIÓN QUE TOMA EL DATAFRAME DE LA TIENDA LUEGO LA  KEY CATEGORÍA DEL PRODUCTO Y AÑADE EN UN DICCIONARIO E INICIA CONTEOS, DEVUELVE
EN DICCIONARIO
'''
def conteo_categoria(tienda):
  diccionario_repeticion={}
  for categoria in tienda["Categoría del Producto"]:
    if categoria not in diccionario_repeticion:
      diccionario_repeticion[categoria]=1
    else:
      diccionario_repeticion[categoria]+=1
  return diccionario_repeticion

#Alternativa más profesional, counter devuelve en un formato símilar al diccionario una subclase de el, se puede convertir usando dict() o trabajar
#En el de la misma manera

#     from collections import Counter

#     diccionario1 = Counter(tienda["Categoría del Producto"])
#     print(diccionario1)
#MUESTRA:
#     Counter({'Muebles': 465, 'Electrónicos': 448, 'Juguetes': 324, 'Electrodomésticos': 312, 'Deportes y diversión': 284, 'Instrumentos musicales': 182, 'Libros': 173, 'Artículos para el hogar': 171})

'''
FUNCIÓN PARA SABER NÚMERO TOTAL DE PEDIDOS
'''
def pedidostotal(ventas_por_categoria):
  return sum(ventas_por_categoria.values())






'''
FUNCIÓN PARA LAS GRÁFICAS DE BARRAS

'''
def grafica_barras(cantidad_colores,x,y,titulo_horizontal,titulo_vertical,titulo):
  plt.figure(figsize=(10, 6))  # Tamaño del gráfico
  colores = plt.cm.Set3(range(len(cantidad_colores)))
  plt.bar(x, y, color=colores)  # Gráfico de barras itera en la misma posición como una tupla
  plt.xlabel(titulo_horizontal)
  plt.ylabel(titulo_vertical)
  plt.title(titulo)
  plt.xticks(rotation=45)  # Rotar etiquetas si son largas
  plt.tight_layout()  # Ajustar espacio automáticamente
  plt.show()





'''
CALIFICACIÓN PROMEDIO PARA LA TIENDA
'''
#utilizo sum con un punto porque funciona en dataframes
def promedio(tienda):
   return round(tienda['Calificación'].sum()/len(tienda['Calificación']),2)




'''
PRODUCTOS VENDIDOS
'''

def conteo_producto(tienda):
  productos_repeticion={}

  for producto in tienda['Producto']:
    if producto not in productos_repeticion:
      productos_repeticion[producto]=1
    else:
      productos_repeticion[producto]+=1
  return productos_repeticion



'''
PRODUCTOS MAX Y MIN VENDIDOS
'''
# Productos más vendidos usamos la funcion max en el diccionario que contiene los productos y su conteo
#Luego recorremos y si la cantidad es igual al mayor obtenido devolvemos una lista con los productos
def maximo(diccionarioproductos_conteo):
  mayor = max(diccionarioproductos_conteo.values())
  productos_mayores = [producto for producto, cantidad in diccionarioproductos_conteo.items() if cantidad == mayor]
  return productos_mayores,mayor

# Productos menos vendidos usamos la función min y values en el diccionario que contiene los productos y su conteo
#PAra saber el valor minimo luego en una lista de compresion recorremos los items del diccionario
#si el valor asociado es igual a menor añadimos los productos que contengas menos ventas
def minimo(diccionarioproducto_conteo):
  menor = min(diccionarioproducto_conteo.values())
  productos_menores = [producto for producto, cantidad in diccionarioproducto_conteo.items() if cantidad == menor]
  return productos_menores,menor

'''
GRÁFICA DE PRODUCTOS MÁS Y MENOS VENDIDOS
'''
def grafica_toma_listas(lista_maximo,lista_minimo,maximo,minimo):
  grafica_conteo_p={}
  for e in lista_maximo:
    grafica_conteo_p[e]=maximo
  for e in lista_minimo:
    grafica_conteo_p[e]=minimo
  grafica_barras(grafica_conteo_p.keys(),grafica_conteo_p.keys(),grafica_conteo_p.values(),"Productos","Ventas","Productos más vendidos tienda ")



'''
Gráfica geográfica Latitud vs longitud
'''
def grafica_latitud_longitud(tienda):
  latitud=[]
  longitud=[]
  for n in tienda['lat']:
    latitud.append(n)
  for n in tienda['lon']:
    longitud.append(n)
  plt.scatter(latitud,longitud)
  plt.title("Latitud vs Longitud")
  plt.xlabel("Latitud")
  plt.ylabel("Longitud")
  plt.show()
````
